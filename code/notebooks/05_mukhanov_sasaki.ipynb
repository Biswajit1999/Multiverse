# Mukhanov-Sasaki numerical benchmark
Integrate the de-Sitter mode equation and compare it against the analytic Bunch-Davies solution.

```python
import sys
from pathlib import Path
root=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src'/'multiverse_cosmology').exists()),None)
sys.path.insert(0,str(root/'src'))
import numpy as np
import matplotlib.pyplot as plt
from multiverse_cosmology.perturbations import solve_de_sitter_mode,de_sitter_mode_analytic
```

```python
eta,v=solve_de_sitter_mode(k=1,eta_i=-80,eta_f=-.08,n=1600)
va=de_sitter_mode_analytic(1,eta)
print('late-time relative amplitude error =',abs(abs(v[-1])-abs(va[-1]))/abs(va[-1]))
plt.loglog(-eta,abs(v),label='numerical')
plt.loglog(-eta,abs(va),'--',label='analytic')
plt.gca().invert_xaxis(); plt.xlabel('-eta'); plt.ylabel('|v_k|'); plt.legend(); plt.grid(alpha=.2); plt.show()
```

This is a benchmark of the perturbation integrator. A later phase will feed arbitrary numerically evolved backgrounds into the mode equation.
