# Vacuum decay and gravitational instanton diagnostics
Explore a metastable scalar potential, thin-wall estimates, the Hawking–Moss exponent and the coupled Euclidean scalar–gravity equations.

```python
import sys
from pathlib import Path
root=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src'/'multiverse_cosmology').exists()),None)
sys.path.insert(0,str(root/'src'))
import numpy as np
import matplotlib.pyplot as plt
from multiverse_cosmology.vacuum_decay import *
```

```python
pts=extrema_tilted_double_well()
true,top,false=identify_vacua(pts)
V=lambda x: tilted_double_well(x)
sigma=wall_tension(V,true.phi,false.phi,true.energy)
R,B=flat_thin_wall_estimate(sigma,false.energy-true.energy)
Bhm=hawking_moss_exponent(false.energy,top.energy)
print(true,top,false)
print('sigma=',sigma,'R_tw=',R,'B_tw=',B,'B_HM=',Bhm)
```

```python
phi=np.linspace(-1.6,1.6,1000)
plt.plot(phi,V(phi)); plt.scatter([true.phi,top.phi,false.phi],[true.energy,top.energy,false.energy]); plt.xlabel('phi'); plt.ylabel('V'); plt.grid(alpha=.2); plt.show()
```

The numerical values belong to a transparent toy potential. They illustrate the vacuum-decay machinery and must not be interpreted as a measured decay rate of our universe.
