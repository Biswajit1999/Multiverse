# Horizon and causal-scale engine
Compute conformal time, particle horizon, event horizon, and the comoving Hubble radius for a baseline FLRW cosmology.

```python
import sys
from pathlib import Path
root=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src'/'multiverse_cosmology').exists()),None)
sys.path.insert(0,str(root/'src'))
import numpy as np
import matplotlib.pyplot as plt
from multiverse_cosmology.friedmann import E
from multiverse_cosmology.horizons import particle_horizon, event_horizon, comoving_hubble_radius
```

```python
a=np.geomspace(1e-8,10,1400)
chi_p=particle_horizon(a,E)
r_h=comoving_hubble_radius(a,E)
plt.loglog(a,np.maximum(chi_p,1e-12),label='particle horizon')
plt.loglog(a,r_h,label='(aH)^-1')
plt.xlabel('scale factor a'); plt.ylabel('H0^-1'); plt.legend(); plt.grid(alpha=.2); plt.show()
```

A horizon is a causal scale, not a physical wall. The calculation makes the room/walls analogy precise by replacing walls with integrals over the spacetime expansion history.
