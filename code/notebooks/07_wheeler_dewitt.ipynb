# Wheeler–DeWitt minisuperspace experiment
Solve a dimensionless closed de-Sitter Wheeler–DeWitt toy equation and compare the numerical barrier with the analytic WKB integral.

```python
import sys
from pathlib import Path
root=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src'/'multiverse_cosmology').exists()),None)
sys.path.insert(0,str(root/'src'))
import numpy as np
import matplotlib.pyplot as plt
from multiverse_cosmology.quantum_cosmology import *
```

```python
lam=.2
print('turning point =',turning_point(lam))
print('WKB action numerical =',wkb_barrier_action(lam),' analytic =',1/(3*lam))
print('log weights =',semiclassical_log_weights(lam))
```

```python
a,psi=solve_wdw(lam,n=1200)
plt.plot(a,psi); plt.axvline(turning_point(lam),ls='--'); plt.xlabel('a'); plt.ylabel('psi(a)'); plt.grid(alpha=.2); plt.show()
```

This is minisuperspace quantum cosmology, not a complete quantum theory of spacetime. The no-boundary and tunnelling labels refer to boundary prescriptions whose interpretation remains an active research topic.
