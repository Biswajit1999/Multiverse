# Inflation parameter sweep and model discrimination
Compute first-order slow-roll predictions for quadratic and Starobinsky-type potentials across a range of e-fold numbers.

```python
import sys
from pathlib import Path
root=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src'/'multiverse_cosmology').exists()),None)
sys.path.insert(0,str(root/'src'))
import numpy as np
import matplotlib.pyplot as plt
from multiverse_cosmology import potentials
from multiverse_cosmology.slowroll import model_curve
```

```python
N=np.linspace(45,65,81)
_,ns_q,r_q=model_curve(N,potentials.quadratic,potentials.d_quadratic,potentials.dd_quadratic,(.2,3),(1.42,30))
_,ns_s,r_s=model_curve(N,potentials.starobinsky,potentials.d_starobinsky,potentials.dd_starobinsky,(.05,3),(1,10))
plt.axvspan(.9649-.0042,.9649+.0042,alpha=.15,label='Planck 2018 n_s baseline')
plt.axhspan(0,.036,alpha=.1,label='BK18 r<0.036')
plt.plot(ns_q,r_q,label='quadratic')
plt.plot(ns_s,r_s,label='Starobinsky')
plt.xlabel('$n_s$'); plt.ylabel('$r$'); plt.xlim(.94,.98); plt.ylim(0,.18); plt.legend(); plt.grid(alpha=.2); plt.show()
```

The observational bands are documented benchmark constraints, not a full 2026 likelihood analysis. The purpose is to show how a concrete model can be tested rather than merely described.
