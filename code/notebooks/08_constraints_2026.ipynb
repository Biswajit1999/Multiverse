# 2026 inflation-data comparison
Compare simple inflationary predictions with versioned compressed constraints from current CMB and CMB+DESI analyses.

```python
import sys
from pathlib import Path
root=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src'/'multiverse_cosmology').exists()),None)
sys.path.insert(0,str(root/'src'))
import numpy as np
import matplotlib.pyplot as plt
from multiverse_cosmology.constraints import *
from multiverse_cosmology import potentials
from multiverse_cosmology.slowroll import model_curve
```

```python
N=np.linspace(45,65,100)
_,ns_q,r_q=model_curve(N,potentials.quadratic,potentials.d_quadratic,potentials.dd_quadratic,(.2,3),(1.42,30))
_,ns_s,r_s=model_curve(N,potentials.starobinsky,potentials.d_starobinsky,potentials.dd_starobinsky,(.05,3),(1,10))
plt.axvspan(CMB_END_2025.ns_mean-CMB_END_2025.ns_sigma,CMB_END_2025.ns_mean+CMB_END_2025.ns_sigma,alpha=.12,label='CMB 1 sigma')
plt.axvspan(CMB_DESI_END_2025.ns_mean-CMB_DESI_END_2025.ns_sigma,CMB_DESI_END_2025.ns_mean+CMB_DESI_END_2025.ns_sigma,alpha=.12,label='CMB+DESI 1 sigma')
plt.axhspan(0,CMB_END_2025.r95_upper,alpha=.08,label='r upper benchmark')
plt.plot(ns_q,r_q,label='quadratic'); plt.plot(ns_s,r_s,label='Starobinsky'); plt.xlabel('ns'); plt.ylabel('r'); plt.legend(); plt.grid(alpha=.2); plt.show()
```

Compressed constraints are useful for transparent diagnostics but do not replace full experiment likelihoods, covariance matrices, foreground/nuisance parameters or model-dependent parameter degeneracies.
