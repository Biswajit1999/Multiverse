# Friedmann scale-factor evolution
Numerically integrate a dimensionless FLRW background and plot $a(\tau)$. 

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
import numpy as np
import matplotlib.pyplot as plt
from src.friedmann import solve_scale_factor
```

```python
sol = solve_scale_factor(t_span=(1e-6, 3.0), a0=1e-4)
t = np.linspace(sol.t[0], sol.t[-1], 1000)
a = sol.sol(t)[0]
plt.figure(figsize=(8,5))
plt.plot(t,a)
plt.xlabel(r'$\tau = H_0 t$')
plt.ylabel('a(t)')
plt.title('FLRW scale-factor evolution')
plt.grid(alpha=0.2)
plt.show()
```

**Interpretation:** this is a background FLRW calculation, not a simulation of an observable external multiverse.
