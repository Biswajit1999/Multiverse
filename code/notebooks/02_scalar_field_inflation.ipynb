# Toy scalar-field inflation
Integrate a homogeneous canonical scalar field in reduced Planck units.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
import numpy as np
import matplotlib.pyplot as plt
from src.inflation import solve_inflation
```

```python
sol = solve_inflation(phi0=15.0, t_end=2e6)
t=sol.t
phi=sol.y[0]
loga=sol.y[2]
fig=plt.figure(figsize=(8,5))
plt.plot(t, phi)
plt.xlabel('t [reduced Planck units]')
plt.ylabel(r'$\phi$')
plt.title('Toy inflaton evolution')
plt.grid(alpha=0.2)
plt.show()

plt.figure(figsize=(8,5))
plt.plot(t, loga-loga[0])
plt.xlabel('t [reduced Planck units]')
plt.ylabel(r'$N = \ln(a/a_0)$')
plt.title('Accumulated expansion')
plt.grid(alpha=0.2)
plt.show()
```

The quadratic potential here is pedagogical. It is not a claim that this is the correct inflaton potential.
